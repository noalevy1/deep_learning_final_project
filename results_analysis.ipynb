# Results & Analysis

In this project, I progressively improved a CNN-based classification system through:

1. Training from scratch
2. Adding regularization (Dropout, BatchNorm, Weight Decay)
3. Strong data augmentation
4. Pretraining on CIFAR-10
5. Transfer learning with ResNet50 (frozen backbone)
6. Fine-tuning ResNet50 (with and without augmentation)

The analysis below focuses on training dynamics, quantitative performance, qualitative observations, and reasoning behind the results.

## Step 1 – Training from Scratch

The SimpleCNN trained with Adam (lr=0.001) converged quickly and smoothly.

### Training & Validation Curves
- Loss decreased sharply in the first few epochs.
- Training and validation curves closely overlapped.
- No visible overfitting.

### Quantitative Results
- Test Accuracy ≈ 93.3%
- Test Loss ≈ 0.29

### Qualitative Observations
The architecture was already well-suited for the dataset.
The small train–validation gap indicates good generalization even without explicit regularization.

### Conclusion
The baseline model was strong, stable, and sufficient for high performance.

## Step 2 – Dropout and Batch Normalization

Adding Dropout (p=0.2) did not change final accuracy, but improved training stability.

Adding Batch Normalization significantly improved convergence speed.

### Training Behavior
- Extremely fast convergence.
- Training and validation accuracy reached ~100%.
- Curves overlapped almost perfectly.

### Test Performance
- Test Accuracy = 100%
- Test Loss ≈ 0

### Reasoning
BatchNorm stabilized gradients and accelerated learning.
Dropout improved stability but did not significantly affect final accuracy.

### Conclusion
Architecture + BatchNorm was the most impactful architectural improvement.


## Step 3 – Regularization Analysis

### Dropout (0.0 vs 0.2 vs 0.5)

All achieved 100% test accuracy.

However:

- Dropout = 0.0 → Validation loss spikes (unstable)
- Dropout = 0.2 → Most stable curves
- Dropout = 0.5 → Slightly slower convergence

Insight:
Regularization improved stability, not final performance.
The dataset is relatively easy for this architecture.

---

### Weight Decay (0.0 vs 1e-4 vs 1e-3)

All achieved 100% test accuracy.

However:
- wd=0.0 → Large validation spikes
- wd=1e-4 → Best stability
- wd=1e-3 → Slightly stronger regularization, still stable

Insight:
Weight decay mainly improved validation stability rather than test accuracy.


## Step 3 – Strong Data Augmentation

Adding strong augmentation increased training difficulty.

### Training Behavior
- Lower early validation accuracy.
- Higher variance in validation loss.
- Converged to high performance (≈96–100%).

### Test Results
- Test Accuracy ≈ 96.7%

### Reasoning
Augmentation forces the model to learn more robust features.
However, for this relatively simple dataset, augmentation slightly reduced peak validation performance.

### Conclusion
Augmentation increased robustness but did not significantly improve final accuracy.


## Step 4 – Pretraining on CIFAR-10

The model learned general visual features:
- Accuracy improved from ~28% to ~56%.
- Saved weights for transfer.

---

## Step 5 – Fine-tuning from CIFAR

Fine-tuning converged very quickly:
- Validation accuracy reached 97% by epoch 3.
- Best validation accuracy = 100%.
- Test accuracy ≈ 93.3%.

### Reasoning
Pretraining accelerated convergence.
However, CIFAR features were less aligned with our binary dataset compared to ImageNet features later.

### Conclusion
Transfer learning from CIFAR helped training speed, but did not outperform stronger architectures.


## Step 6 – ResNet50 (Frozen Backbone)

Using ImageNet-pretrained ResNet50 with frozen backbone:

### Training Behavior
- Extremely fast convergence.
- Smooth and stable curves.
- Validation accuracy reached 100% by epoch 4.

### Test Performance
- Test Accuracy = 100%

### Reasoning
ImageNet features are highly transferable.
Even without fine-tuning, the representation was strong enough.

### Conclusion
Transfer learning significantly improved efficiency and stability.


## Step 7 – ResNet50 Fine-Tuning

### Without Augmentation
- Validation accuracy = 100%
- Test accuracy = 100%
- Extremely fast convergence

### With Strong Augmentation
- Validation accuracy ≈ 96.7%
- Test accuracy = 100%
- Slower convergence

### Reasoning
For this small and simple dataset, ResNet50 is already very powerful.
Strong augmentation made optimization harder without improving generalization.

### Final Insight
Transfer learning was the most impactful improvement in the entire project.


# Final Reflection

### What architectural choices mattered most?

Batch Normalization had the strongest architectural impact.
It improved convergence speed and stability dramatically.

Switching from a custom CNN to a pretrained ResNet50 was the most significant improvement overall.

---

### Which optimization strategies worked best?

Adam optimizer consistently produced fast and stable convergence.

Moderate regularization (dropout=0.2, wd=1e-4) provided the best balance between stability and performance.

---

### When did transfer learning help?

Transfer learning helped in two ways:
1. CIFAR pretraining accelerated convergence.
2. ImageNet-pretrained ResNet50 provided near-perfect performance immediately.

The ImageNet transfer was substantially more impactful than CIFAR pretraining.

---

### What would I do differently with more time or data?

- Use a larger and more challenging dataset.
- Evaluate cross-validation instead of a single split.
- Perform hyperparameter search (learning rate scheduling, cosine decay).
- Compare additional architectures (EfficientNet, ViT).
- Analyze feature embeddings visually (t-SNE).

---

### Overall Conclusion

The dataset is relatively simple for modern CNN architectures.
Regularization improved stability, not final accuracy.
Transfer learning was the most powerful factor in achieving fast, stable, and perfect performance.
